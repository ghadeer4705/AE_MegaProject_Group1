```python
import yaml
yaml_path = '/content/dataset/data.yaml'

with open(yaml_path, 'r') as f:
    data = yaml.safe_load(f)

data['path'] = '/content/dataset'

with open(yaml_path, 'w') as f:
    yaml.dump(data, f)
print(data)
```

```python
from ultralytics import YOLO
model=YOLO('yolo26n.pt')

results=model.train(
    data='/content/dataset/data.yaml',
    epochs=50,
    imgsz=640,
    batch=16,
    device=0,
    project='yolo26_run',
    name='exp1'
)
```

```python
model=YOLO('/content/runs/detect/yolo26_run/exp1/weights/best.pt')
test_results=model.val(data='/content/dataset/data.yaml',split='test')
print("Evaluation dataset results:")
print(f"Precision: {test_results.results_dict['metrics/precision(B)']:.4f}")
print(f"Recall:    {test_results.results_dict['metrics/recall(B)']:.4f}")
print(f"mAP50:     {test_results.results_dict['metrics/mAP50(B)']:.4f}")
print(f"mAP50-95:  {test_results.results_dict['metrics/mAP50-95(B)']:.4f}")
```
